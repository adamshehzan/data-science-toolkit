```python
from google.colab import files
import pandas as pd

file_name = 'train_and_test2.csv'
df = pd.read_csv(file_name)

print("✅ Dataset loaded successfully!")
print("Shape:", df.shape)

print("=== First 5 rows ===")
display(df.head())
```

```python
total_samples = df.shape[0]
print(f"Total number of samples/rows: {total_samples}")

if total_samples < 100:
    print("⚠️ The dataset has very few samples. It may NOT be enough for reliable analysis/ML.")
elif total_samples < 1000:
    print("⚠️ The dataset is relatively small. It may be enough for basic analysis but limited for ML.")
elif total_samples < 10000:
    print("✅ The dataset has a moderate number of samples. Generally sufficient for most tasks.")
else:
    print("✅ The dataset has a large number of samples. More than enough for analysis and ML.")
```

```python
num_features = df.shape[1]
print(f"Total number of features/columns: {num_features}")
print(f"\nFeature names: {list(df.columns)}")

# Evaluate
if num_features < 3:
    print("⚠️ Very few features. May not capture enough information.")
elif num_features < 10:
    print("✅ Moderate number of features. Usually sufficient for basic modeling.")
elif num_features < 50:
    print("✅ Good number of features. Plenty of information for modeling.")
else:
    print("⚠️ Many features. Consider feature selection to avoid overfitting/curse of dimensionality.")
```

```python
num_samples = df.shape[0]
num_features = df.shape[1]

print(f"Number of samples (observations): {num_samples}")
print(f"Number of features: {num_features}")
print(f"Sample-to-feature ratio: {num_samples / num_features:.2f}")

ratio = num_samples / num_features
if ratio < 10:
    print(f"⚠️ Ratio is {ratio:.2f}. You may need more samples (aim for at least 10x features).")
else:
    print(f"✅ Ratio is {ratio:.2f}. Sample size is sufficient relative to number of features.")
```

```python
print("=== Datatypes of each column ===")
print(df.dtypes)

print("\n=== Summary ===")
print(f"Numerical columns: {df.select_dtypes(include=['int64', 'float64']).shape[1]}")
print(f"Categorical columns: {df.select_dtypes(include=['object', 'category']).shape[1]}")
print(f"Boolean columns: {df.select_dtypes(include=['bool']).shape[1]}")
print(f"DateTime columns: {df.select_dtypes(include=['datetime']).shape[1]}")
```

```python
null_counts = df.isnull().sum()
null_cols = null_counts[null_counts > 0]

print(f"Total columns with null values: {len(null_cols)}")

if len(null_cols) > 0:
    print("\n=== Columns with Null Values ===")
    for col, count in null_cols.items():
        rows_with_null = df[df[col].isnull()].index.tolist()
        print(f"Column '{col}': {count} null values")
        print(f"  → Rows with nulls: {rows_with_null[:10]}{'...' if len(rows_with_null) > 10 else ''}")
else:
    print("✅ No null values found in any column.")

```

```python
missing = df.isnull()
total_missing = missing.sum().sum()
print(f"Total missing values in dataset: {total_missing}")

print("\n=== Missing Values per Column ===")
for col in df.columns:
    count = missing[col].sum()
    if count > 0:
        pct = (count / len(df)) * 100
        rows = df[missing[col]].index.tolist()
        print(f"Column '{col}': {count} missing ({pct:.2f}%)")
        print(f"  → Row indices: {rows[:15]}{'...' if len(rows) > 15 else ''}")

# Summary table
missing_df = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': df.isnull().sum().values,
    'Missing_Percent': (df.isnull().sum().values / len(df) * 100).round(2)
})
missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)
display(missing_df)
```
