```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones: ")
print(ones_array)

print("Array of Zeros: ")
print(zeros_array)
```
