```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones: ")
print(ones_array)

print("Array of Zeros: ")
print(zeros_array)
```

```python
import numpy as np

arr4 = np.random.rand(6, 3)
arr4_transpose = arr4.T

print("4. Random 6x3 Array:")
print(arr4)
print("   Transposed (3x6):")
print(arr4_transpose)
```

```python
n, p, m = 4, 5, 3
arr5_a = np.random.rand(n, p)
arr5_b = np.random.rand(p, m)
dot_product = np.dot(arr5_a, arr5_b)

print(f"5. Array A ({n}x{p}):")
print(arr5_a)
print(f"   Array B ({p}x{m}):")
print(arr5_b)
print(f"   Dot Product ({n}x{m}):")
print(dot_product)
```
